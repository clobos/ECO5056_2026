# Cria o boxplot

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.boxplot(x='especies', y='comprimento_sepala', data=dados)
plt.title('Boxplot do Comprimento da Sépala por Espécie')
plt.xlabel('Espécie')
plt.ylabel('Comprimento da Sépala (cm)')
plt.show()
```

# Cria o gráfico de dispersão

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(x='comprimento_sepala', y='largura_sepala', hue='especies', data=dados, s = 50)
plt.title('Gráfico de Dispersão: Comprimento vs. Largura da Sépala por Espécie')
plt.xlabel('Comprimento da Sépala (cm)')
plt.ylabel('Largura da Sépala (cm)')
plt.legend(title='Espécie')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=dados, x='comprimento_sepala')
plt.title('Histograma do Comprimento da Sépala por Espécie')
plt.xlabel('Comprimento da Sépala (cm)')
plt.ylabel('Frequência')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=dados, x='comprimento_petala')
plt.title('Histograma do Comprimento da Pétala por Espécie')
plt.xlabel('Comprimento da Pétala (cm)')
plt.ylabel('Frequência')
plt.show()
```

# Cria uma figura e um conjunto de subplots (2 linhas, 2 colunas)

```python

fig, ax = plt.subplots(2, 2, figsize=(12, 6))

# Plot 1: Boxplot do Comprimento da Sépala por Espécie (ax[0, 0])
sns.boxplot(x='especies', y='comprimento_sepala', data=dados, ax=ax[0, 0], color= "yellow")
ax[0, 0].set_title('Boxplot do Comprimento da Sépala por Espécie')
ax[0, 0].set_xlabel('Espécie')
ax[0, 0].set_ylabel('Comprimento da Sépala (cm)')

# Plot 2: Gráfico de Dispersão: Comprimento vs. Largura da Sépala por Espécie (ax[0, 1])
sns.scatterplot(x='comprimento_sepala', y='largura_sepala', hue='especies', data=dados, s=50, ax=ax[0, 1])
ax[0, 1].set_title('Gráfico de Dispersão: Comprimento vs. Largura da Sépala por Espécie')
ax[0, 1].set_xlabel('Comprimento da Sépala (cm)')
ax[0, 1].set_ylabel('Largura da Sépala (cm)')
ax[0, 1].legend(title='Espécie')

# Plot 3: Histograma do Comprimento da Sépala (ax[1, 0])
sns.histplot(data=dados, x='comprimento_sepala', ax=ax[1, 0], color="red")
ax[1, 0].set_title('Histograma do Comprimento da Sépala')
ax[1, 0].set_xlabel('Comprimento da Sépala (cm)')
ax[1, 0].set_ylabel('Frequência')

# Plot 4: Histograma do Comprimento da Pétala (ax[1, 1])
sns.histplot(data=dados, x='comprimento_petala', ax=ax[1, 1])
ax[1, 1].set_title('Histograma do Comprimento da Pétala')
ax[1, 1].set_xlabel('Comprimento da Pétala (cm)')
ax[1, 1].set_ylabel('Frequência')

# Ajusta o layout para evitar sobreposição
plt.tight_layout()

# Exibe a figura combinada
plt.show()
```
